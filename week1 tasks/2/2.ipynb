````python
```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12

df = pd.read_csv('gapminder.csv')

print(f"Data Shape: {df.shape}")
print(df.head())

plt.figure(figsize=(15, 10))
sns.lineplot(data=df, x='year', y='lifeExp', hue='country', legend='full', palette='tab20')

plt.title('Life Expectancy by Country (1952-2007) - The "Spaghetti Plot"', fontsize=16)
plt.xlabel('Year')
plt.ylabel('Life Expectancy (Years)')
plt.legend(title='Country', bbox_to_anchor=(1.05, 1), loc='upper left', ncol=2)
plt.tight_layout()
plt.show()

continent_agg = df.groupby(['year', 'continent'])['lifeExp'].mean().reset_index()

plt.figure(figsize=(12, 8))
sns.lineplot(data=continent_agg, x='year', y='lifeExp', hue='continent', marker='o', linewidth=2.5)

plt.title('Average Life Expectancy by Continent (1952-2007)', fontsize=16)
plt.xlabel('Year')
plt.ylabel('Life Expectancy (Years)')
plt.legend(title='Continent')
plt.tight_layout()
plt.show()

highlight_countries = ['China', 'Sierra Leone', 'Japan']

df['Highlight'] = df['country'].apply(lambda x: x if x in highlight_countries else 'Other')

plt.figure(figsize=(12, 8))

sns.lineplot(
    data=df[df['Highlight'] == 'Other'],
    x='year',
    y='lifeExp',
    color='lightgrey',
    linewidth=1,
    label=None
)

sns.lineplot(
    data=df[df['Highlight'] != 'Other'],
    x='year',
    y='lifeExp',
    hue='country',
    linewidth=3,
    marker='o'
)

last_year = 2007
for country in highlight_countries:
    row = df[(df['country'] == country) & (df['year'] == last_year)]
    if not row.empty:
        plt.text(
            row['year'].values[0] + 0.5,
            row['lifeExp'].values[0],
            country,
            fontsize=12,
            fontweight='bold',
            va='center'
        )

plt.title('Life Expectancy Trends: Selected Countries vs Global Background', fontsize=16)
plt.xlabel('Year')
plt.ylabel('Life Expectancy (Years)')
plt.xlim(1950, 2010)
plt.gca().get_legend().remove()
plt.tight_layout()
plt.show()

df_gap = df[df['year'].isin([1952, 2007])].copy()

stats = df_gap.groupby('year')['gdpPercap'].agg(['min', 'max']).reset_index()

plt.figure(figsize=(10, 6))

years = [1952, 2007]
colors = ['#1f77b4', '#ff7f0e']

for i, year in enumerate(years):
    row = stats[stats['year'] == year].iloc[0]

    plt.plot(
        [row['min'], row['max']],
        [i, i],
        color=colors[i],
        linewidth=4,
        alpha=0.7,
        label=f'{year} Range'
    )

    plt.scatter([row['min'], row['max']], [i, i], color=colors[i], s=100, zorder=5)

    plt.text(row['min'], i - 0.15, f"${int(row['min']):,}", ha='center', fontsize=9)
    plt.text(row['max'], i - 0.15, f"${int(row['max']):,}", ha='center', fontsize=9)

plt.yticks([0, 1], ['1952', '2007'])
plt.xlabel('GDP Per Capita (USD)')
plt.title('Gap Between Richest and Poorest Countries: 1952 vs 2007', fontsize=14)

plt.xscale('log')
plt.grid(True, which="both", ls="-", alpha=0.2)

import matplotlib.ticker as ticker
plt.gca().xaxis.set_major_formatter(ticker.StrMethodFormatter("${x:,.0f}"))

plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```
````

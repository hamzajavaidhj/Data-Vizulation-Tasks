```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12

df = pd.read_csv('penguins.csv')

print(f"Original Shape: {df.shape}")
print(df.head())
print("\nMissing Values Summary:")
print(df.isnull().sum())

measurements_cols = [
    'bill_length_mm',
    'bill_depth_mm',
    'flipper_length_mm',
    'body_mass_g'
]

missing_all_measurements = df[measurements_cols].isnull().all(axis=1)
print(f"Penguins missing all measurements: {missing_all_measurements.sum()}")

missing_sex = df['sex'].isnull()
print(f"Penguins missing sex: {missing_sex.sum()}")

df_clean = df.dropna(subset=measurements_cols).copy()
print(f"\nShape after dropping empty rows: {df_clean.shape}")

df_sex_clean = df_clean.dropna(subset=['sex']).copy()
print(f"Shape after dropping missing sex: {df_sex_clean.shape}")

plt.figure(figsize=(10, 6))
sns.histplot(
    data=df_clean,
    x='body_mass_g',
    bins=30,
    kde=True,
    color='steelblue'
)
plt.title('Distribution of Penguin Body Mass', fontsize=16)
plt.xlabel('Body Mass (g)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.histplot(
    data=df_clean,
    x='body_mass_g',
    hue='species',
    bins=30,
    kde=True,
    palette='Set2'
)
plt.title('Distribution of Body Mass by Species', fontsize=16)
plt.xlabel('Body Mass (g)')
plt.ylabel('Count')
plt.legend(title='Species')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 8))
sns.scatterplot(
    data=df_clean,
    x='bill_length_mm',
    y='bill_depth_mm',
    hue='species',
    palette='Set2',
    s=100,
    alpha=0.7
)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(10, 8))

sns.scatterplot(
    data=df_clean,
    x='bill_length_mm',
    y='bill_depth_mm',
    hue='species',
    palette='Set2',
    s=100,
    alpha=0.7,
    ax=ax
)

for species in df_clean['species'].unique():
    species_data = df_clean[df_clean['species'] == species]

    coefficients = np.polyfit(
        species_data['bill_length_mm'],
        species_data['bill_depth_mm'],
        1
    )

    regression = np.poly1d(coefficients)

    x_values = np.linspace(
        species_data['bill_length_mm'].min(),
        species_data['bill_length_mm'].max(),
        100
    )

    ax.plot(
        x_values,
        regression(x_values),
        linestyle='--',
        linewidth=2,
        label=f'{species} Trend'
    )

plt.title('Bill Length vs. Bill Depth by Species', fontsize=16)
plt.xlabel('Bill Length (mm)')
plt.ylabel('Bill Depth (mm)')
plt.legend(title='Species/Trend')
plt.tight_layout()
plt.show()

avg_mass = (
    df_sex_clean
    .groupby(['species', 'sex'])['body_mass_g']
    .mean()
    .reset_index()
)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=avg_mass,
    x='species',
    y='body_mass_g',
    hue='sex',
    palette='viridis',
    errorbar=None
)

plt.title('Average Body Mass by Species and Sex', fontsize=16)
plt.xlabel('Species')
plt.ylabel('Average Body Mass (g)')
plt.legend(title='Sex')
plt.tight_layout()
plt.show()
```
